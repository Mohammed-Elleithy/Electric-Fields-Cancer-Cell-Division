```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time = np.linspace(0,10,1000)

field = np.sin(20*time)

plt.plot(time,field)
plt.title("Alternating Electric Field")
plt.xlabel("Time")
plt.ylabel("Field Strength")
plt.show()
```

```python
baseline_growth = np.exp(time/10)

ttf_growth = baseline_growth * 0.6

plt.plot(time,baseline_growth,label="No TTFields")
plt.plot(time,ttf_growth,label="TTFields Applied")
plt.legend()
plt.show()
```

**Interpretation**

This is a conceptual visualization showing how TTFields attempt to reduce tumor proliferation.

**Real TTFields disrupt:**

mitotic spindle formation,
chromosome segregation,
cytokinesis.
